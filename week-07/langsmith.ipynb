```python
import os

# --- LangSmith 설정 ---
os.environ["LANGSMITH_TRACING"] = "true"
os.environ["LANGSMITH_ENDPOINT"] = "https://apac.api.smith.langchain.com"
os.environ["LANGSMITH_PROJECT"] = "adapterz-langchain-textbook"

# LANGSMITH_API_KEY와 GOOGLE_API_KEY는 실행 전에 환경 변수로 설정합니다.

```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

chain = (
    ChatPromptTemplate.from_messages([
        ("system", "당신은 친절한 한국어 비서입니다."),
        ("human", "{question}"),
    ])
    | ChatGoogleGenerativeAI(
        model="gemini-2.5-flash-lite",
        api_key=os.getenv("GOOGLE_API_KEY") or os.environ["GEMINI_API_KEY"],
    )
    | StrOutputParser()
)

answer = chain.invoke({"question": "LangSmith는 어떤 도구인가요?"})
print(answer)
```
